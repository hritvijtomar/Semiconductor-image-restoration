```python
import os

# Automatically set the root project directory (KLA folder)
current_dir = os.getcwd()
if os.path.basename(current_dir) == "notebooks":
    project_root = os.path.abspath("..")
else:
    project_root = current_dir

# Set absolute directory paths
gt_dir = os.path.join(project_root, "train", "train", "GT")
noisy_dir = os.path.join(project_root, "train", "train", "NoisyLR")
test_dir = os.path.join(project_root, "Test_NoisyLR", "NoisyLR")

# Read valid .npy files (ignoring system files like .DS_Store)
gt_files = set(f for f in os.listdir(gt_dir) if f.endswith('.npy'))
noisy_files = set(f for f in os.listdir(noisy_dir) if f.endswith('.npy'))
test_files = set(f for f in os.listdir(test_dir) if f.endswith('.npy'))

# Verification stats
matching_pairs = len(gt_files.intersection(noisy_files))
missing_in_noisy = len(gt_files - noisy_files)
missing_in_gt = len(noisy_files - gt_files)

# Print output
print(f"GT images: {len(gt_files)}")
print(f"NoisyLR images: {len(noisy_files)}")
print(f"Test images: {len(test_files)}")
print()
print(f"Matching pairs: {matching_pairs}")
print(f"Missing in NoisyLR: {missing_in_noisy}")
print(f"Missing in GT: {missing_in_gt}")
```
